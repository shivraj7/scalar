```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv('final_vg.csv')
```

```python
data
```

Multivariate Analysis


```python
top3_pub = data['Publisher'].value_counts().index[:3]
top3_gen = data['Genre'].value_counts().index[:3]
top3_plat = data['Platform'].value_counts().index[:3]
top3_data = data.loc[(data["Publisher"].isin(top3_pub)) & (data["Platform"].isin(top3_plat)) & (data['Genre'].isin(top3_gen))]
top3_data
```

```python
sns.scatterplot(data=top3_data,x='NA_Sales',y='EU_Sales')
plt.show()
```

```python
sns.scatterplot(data=top3_data,x='NA_Sales',y='EU_Sales', hue='Publisher') # NNC
plt.show()
```

observations

EA sales > activision sales > Namco sales

for both EU and NA


```python
sns.scatterplot(data=top3_data,x='NA_Sales',y='EU_Sales', hue='Genre') # NNC
plt.show()
```

Observations

sports sales > action sales > misc sales

for both EU and NA

```python
top3_data.head()
```

```python
sns.boxplot(data=top3_data, x='Genre', y= 'Global_Sales', )
plt.show()
```

```python
sns.boxplot(data=top3_data, x='Publisher', y= 'Global_Sales', )
plt.show()
```

```python
sns.boxplot(data=top3_data, x='Publisher', y= 'Global_Sales', hue='Genre') #CNN
plt.show()
```

```python
#average global sales for each of my publisher
sns.barplot(data=top3_data, x='Publisher', y='Global_Sales')
```

#joint plot

```python
sns.jointplot(data=top3_data,x='NA_Sales', y='EU_Sales')
```

```python
sns.jointplot(data=top3_data,x='NA_Sales', y='EU_Sales', hue='Genre')
```

```python
sns.jointplot(data=top3_data,x='NA_Sales', y='EU_Sales', kind='reg')
```

```python
sns.jointplot(data=top3_data,x='NA_Sales', y='EU_Sales', kind='kde')
```

```python
sns.jointplot(data=top3_data,x='NA_Sales', y='EU_Sales', kind='hex')
```

#Pair Plot

```python
#this makes scatter plots of pair of numerical data
sns.pairplot(data=top3_data)
plt.show()
```

#Pearson Correlation

```python
#range -1 to +1
top3_data.corr(numeric_only=True)
```

```python
sns.heatmap(top3_data.corr(numeric_only=True))
```

```python
sns.heatmap(top3_data.corr(numeric_only=True), cmap='Blues')
plt.show()
```

```python
sns.heatmap(top3_data.corr(numeric_only=True), cmap='Blues', annot=True)
plt.show()
```
